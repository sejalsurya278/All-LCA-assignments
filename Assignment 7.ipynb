```python
# Assignment 7: Data Visualization

# 1. Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Load the CSV dataset
df = pd.read_csv("students.csv")

# 3. Display the dataset
print("Student Dataset:")
print(df)

# 4. Line Plot for Python Marks
plt.figure(figsize=(8, 5))
plt.plot(df["Student"], df["Python"], marker="o")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.show()

# 5. Bar Plot for Maths Marks
plt.figure(figsize=(8, 5))
plt.bar(df["Student"], df["Maths"])
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.show()

# 6. Scatter Plot for Maths vs Python Marks
plt.figure(figsize=(8, 5))
plt.scatter(df["Maths"], df["Python"])
plt.title("Maths vs Python Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.show()

# 7. Calculate Correlation Matrix
correlation = df[["Maths", "Python", "Attendance"]].corr()

print("Correlation Matrix:")
print(correlation)

# 8. Create Heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
