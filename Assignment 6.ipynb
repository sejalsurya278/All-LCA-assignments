```python

# Assignment 6: Data Preprocessing using Pandas

# 1. Import Pandas
import pandas as pd

# 2. Load the Titanic dataset
df = pd.read_csv("Titanic-dataset.csv")

# 3. Display the dataset
print("Original Dataset:")
print(df.head())

# 4. Explore the dataset
print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Description:")
print(df.describe())

# 5. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 6. Handle missing values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# 7. Identify duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())

# 8. Remove duplicate records
df.drop_duplicates(inplace=True)

# 9. Remove irrelevant column
df.drop("Cabin", axis=1, inplace=True)

# 10. Select specific columns using loc
print("\nSelected Columns:")
print(df.loc[:, ["Name", "Age", "Sex"]])

# 11. Filter passengers older than 30
print("\nPassengers Older Than 30:")
print(df.loc[df["Age"] > 30, ["Name", "Age", "Sex"]])

# 12. Display cleaned dataset
print("\nCleaned Dataset:")
print(df.head())

print("\nFinal Dataset Shape:")
print(df.shape)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```
