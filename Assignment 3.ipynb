```python
#Assignment 3- whether the triangle is right angled triangle or not

a=int(input("enter 1st side of triangle"))
b=int(input("enter 2nd side of triangle"))
c=int(input("enter 3rd side of triangle"))
      
def right_triangle(a,b,c):
 if a*a + b*b == c*c:
      print("Triangle is right angle triangle")
 else:
      print("triangle is not a right angled triangle")

right_triangle(a,b,c)

```
